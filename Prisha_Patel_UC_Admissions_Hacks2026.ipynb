```python
#q5

eth_2025 = ethnicity[
    (ethnicity["fall_term"] == 2025) &
    (ethnicity["entrant_level"] == "freshman")
].copy()

# Applicants
app = eth_2025[
    eth_2025["count_type"] == "App"
]

# Admits
adm = eth_2025[
    eth_2025["count_type"] == "Adm"
]

# White
white_app = app[
    app["ethnicity"] == "White"
].groupby("campus")["n"].sum()

white_adm = adm[
    adm["ethnicity"] == "White"
].groupby("campus")["n"].sum()

# Hispanic/Latino(a)
hisp_app = app[
    app["ethnicity"] == "Hispanic/Latino(a)"
].groupby("campus")["n"].sum()

hisp_adm = adm[
    adm["ethnicity"] == "Hispanic/Latino(a)"
].groupby("campus")["n"].sum()

# Admit rates
white_rate = white_adm / white_app * 100
hisp_rate = hisp_adm / hisp_app * 100

# Compare
comparison = pd.DataFrame({
    "White Admit Rate": white_rate,
    "Hispanic/Latino(a) Admit Rate": hisp_rate
})

comparison["White Higher"] = (
    comparison["White Admit Rate"] >
    comparison["Hispanic/Latino(a) Admit Rate"]
)

print(comparison)

print(
    "\nNumber of campuses where White rate is higher:",
    comparison["White Higher"].sum()
)
```

```python
#Q5

eth_2025 = ethnicity[
    (ethnicity["fall_term"] == 2025) &
    (ethnicity["entrant_level"] == "freshman") &
    (ethnicity["campus"] != "Systemwide")
].copy()

# Separate applicants and admits
app = eth_2025[eth_2025["count_type"] == "App"]
adm = eth_2025[eth_2025["count_type"] == "Adm"]

# White
white_app = app[
    app["ethnicity"] == "White"
].set_index("campus")["n"]

white_adm = adm[
    adm["ethnicity"] == "White"
].set_index("campus")["n"]

# Hispanic/Latino(a)
hisp_app = app[
    app["ethnicity"] == "Hispanic/Latino(a)"
].set_index("campus")["n"]

hisp_adm = adm[
    adm["ethnicity"] == "Hispanic/Latino(a)"
].set_index("campus")["n"]

# Calculate admit rates
white_rate = white_adm / white_app * 100
hisp_rate = hisp_adm / hisp_app * 100

# Compare
comparison = pd.DataFrame({
    "White Admit Rate": white_rate,
    "Hispanic/Latino(a) Admit Rate": hisp_rate
})

comparison["White Higher"] = (
    comparison["White Admit Rate"] >
    comparison["Hispanic/Latino(a) Admit Rate"]
)

print(comparison.round(2))

print(
    "\nNumber of campuses where White rate is higher:",
    int(comparison["White Higher"].sum())
)
```

```python
#Q7

df = pd.read_csv(
    "uc_data/bay_area_modeling_table.csv",
    low_memory=False
)

bay_area_counties = [
    "Alameda",
    "Contra Costa",
    "Marin",
    "Napa",
    "San Francisco",
    "San Mateo",
    "Santa Clara",
    "Solano",
    "Sonoma"
]

bay_2023 = df[
    (df["fall_term"] == 2023) &
    (df["campus"] == "Universitywide") &
    (df["county"].isin(bay_area_counties))
].copy()

print("Number of schools:", bay_2023["high_school"].nunique())
print("Graduates:", bay_2023["graduates"].sum())
print("CCC enrollment:", bay_2023["enrolled_ccc"].sum())
```

```python
graduates = bay_2023["graduates"].sum()
ccc = bay_2023["enrolled_ccc"].sum()

share = ccc / graduates * 100

print("CCC enrollment share:", round(share, 2), "%")
```
